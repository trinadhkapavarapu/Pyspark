**🔹 greatest(cols)**

- •	Definition: Returns the largest value among the specified columns for each row.
- •	Type: Row-wise comparison.
- •	Example: greatest("col1", "col2", "col3") returns the highest value among those columns for each row.

**🔹 max(col)**

- •	Definition: Returns the maximum value of a column across all rows.
- •	Type: Aggregate function.
- •	Example: df.agg(max("col1")) returns the highest value in column col1 over the entire DataFrame.

**🔹 least(cols)**
- •	Definition: Returns the smallest value among the specified columns for each row.
- •	Type: Row-wise comparison.
- •	Example: least("col1", "col2", "col3") returns the lowest value among those columns for each row.

**🔹 min(col)**

- •	Definition: Returns the minimum value of a column across all rows.
- •	Type: Aggregate function.
- •	Example: df.agg(min("col1")) returns the lowest value in column col1 over the entire DataFrame.

```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import greatest, least, max, min
# Sample data
input_data = [
    ("David", 70, 68, 89, 40, 84),
    ("Kevin", 90, 67, 87, 79, 74),
    ("Natalia", 66, 88, 49, 65, 72),
    ("Roger", 68, 73, 82, 59, 67),
    ("Michael", 80, 86, 69, 78, 92)
]
schema = ["Student", "Subject_1", "Subject_2", "Subject_3", "Subject_4", "Subject_5"]
# Create DataFrame
df = spark.createDataFrame(input_data, schema)
# Add row-wise greatest and least columns
df_with_extremes = df.withColumn("Greatest", greatest("Subject_1", "Subject_2", "Subject_3", "Subject_4", "Subject_5")) \
                     .withColumn("Least", least("Subject_1", "Subject_2", "Subject_3", "Subject_4", "Subject_5"))
# Show updated DataFrame
df_with_extremes.show()
# Aggregate max and min for Subject_1
df.agg(
    max("Subject_1").alias("Max_Subject_1"),
    min("Subject_1").alias("Min_Subject_1")
).show()
```
