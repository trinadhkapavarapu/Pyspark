**write data frame into parquet file using pyspark**

```python
data = [(1, 'jhansi'), (2, 'jhanu')]
schema = ['id', 'name']
df = spark.createDataFrame(data=data, schema=schema)

df.write.parquet(path='/Volumes/tri/tri/tri_files/OUTPUT/', mode='overwrite')
display(spark.read.parquet('/Volumes/tri/tri/tri_files/OUTPUT/'))
```

**we can use mode = ignore /overwrite/append/error**

```python

df.write. parquet (path='dbfs:/tmp/emps',header=true,mode = 'error')#-data there it gives
df.write. parquet (path='dbfs:/tmp/emps',header=true,mode = 'overwrite')#-overwrite existing data
df.write. parquet (path='dbfs:/tmp/emps',header=true,mode = 'ignore')#-throw error data there
```

note: header=True is not applicable for Parquet files.
