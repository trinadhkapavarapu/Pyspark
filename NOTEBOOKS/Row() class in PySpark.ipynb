**Row()
Definition:**

pyspark.sql.Row() is used to represent a single record (row) in a DataFrame. It allows you to create row objects using positional values or named arguments.

# #✅ Example 1 — Row by position (index access)

```python
#✅ Example 1 — Row by position (index access)
from pyspark.sql import Row

r = Row('tri', 1000) # created by position
print(r[0] + ' ' + str(r[1])) # tri 1000



```

#Notes
- •	When you create a Row by position, you access fields by index (r[0], r[1]).
- •	There are no names on the fields here, so r.name won’t work.

# #✅ Example 2 — Row by named arguments (name access)

```python
#✅ Example 2 — Row by named arguments (name access)
from pyspark.sql import Row

r = Row(name='tri', salary=1000) # created by name
# Two equivalent ways to access:
print(r['name'] + ' ' + str(r['salary'])) # tri 1000
print(r.name + ' ' + str(r.salary)) # tri 1000

#Tip: Use either dict-style (r['name']) or attribute-style (r.name) for named rows.
```

# ✅ Example 3 — Row factory (Row-like “class”) and DataFrame creation

```python
from pyspark.sql import Row

# Create a Row *factory* with field names
Person = Row('name', 'salary')

# Create row instances
p1 = Person('maheer', 2000)
p2 = Person('tri', 10000)

print(p1.name + ' ' + p2.name) # maheer tri

# Build a DataFrame from a list of Rows
data = [p1, p2]
df = spark.createDataFrame(data)
df.show()
df.printSchema()

```

# ✅ Example 4 — Nested Row (struct field in a DataFrame)

```python
from pyspark.sql import Row

data = [
Row(name='trinadh', prop=Row(age=30, gender='male')),
Row(name='nadh', prop=Row(age=35, gender='male'))
]
df = spark.createDataFrame(data)

df.show(truncate=False)
df.printSchema()

# Access nested fields
df.select('name', 'prop.age', 'prop.gender').show()

```

# ✅ Example 5 Bonus: Create DataFrame directly from named Rows

```python
from pyspark.sql import Row

rows = [
Row(name='tri', salary=1000),
Row(name='sri', salary=1200),
]
person_df = spark.createDataFrame(rows)
person_df.show()
```
