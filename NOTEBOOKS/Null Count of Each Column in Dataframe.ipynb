```python
#dataframe
data_student = [
    ("Michael", "Science", 80, "P", 90),
    ("Nancy", "Mathematics", 90, "P", None),
    ("David", "English", 20, "F", 80),
    ("John", "Science", None, "F", None),
    ("Martin", "Mathematics", None, None, 70),
    (None, None, None, None, None)
]
Schema = ["name", "Subject", "Mark", "Status", "Attendance"]
df = spark.createDataFrame(data=data_student, schema=Schema)
display(df)
```

```python
#find null occurrences of each column in dataframe
from pyspark.sql.functions import col, count, when
result = df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns])
display(result)
```

****note🔹 Explanation**
- •	df.columns: Gets a list of all column names in the DataFrame.
- •	col(c).isNull(): Checks if the value in column c is null.
- •	when(condition, value): Returns value when condition is true; otherwise returns null.
- •	count(...): Counts the number of non-null values returned by when(...), which effectively counts the number of nulls in each column.
- •	alias(c): Renames the resulting column to match the original column name.
