**🧠 When to Use json_tuple() in PySpark**
The json_tuple() function is used to extract multiple fields from a JSON string column in PySpark without defining a schema. It’s simple and efficient for flat JSON structures.
________________________________________
**✅ Use json_tuple() When:**
- 1.	You have a flat JSON string (no nested objects or arrays).
- 2.	You want to extract multiple fields quickly without using from_json() or get_json_object().
- 3.	You don’t need to define a schema — it returns each field as a separate column.
- 4.	You want better performance for simple JSON parsing.


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import json_tuple
# Create Spark session
spark = SparkSession.builder.getOrCreate()
# Correct JSON strings (double quotes)
data = [
    ('trinadh', '{"hair":"black","eye":"brown","skin":"black"}'),
    ('nadh', '{"hair":"black","eye":"brown","skin":"white"}')
]
schema = ['name', 'props']
# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show(truncate=False)
df.printSchema()

```

# Extract values using `json_tuple`

```python
# Extract values using json_tuple
df2 = df.select(
    df.name,
    json_tuple(df.props, 'hair', 'skin').alias('hair', 'skin')
)
df2.show()
df.printSchema()
```
