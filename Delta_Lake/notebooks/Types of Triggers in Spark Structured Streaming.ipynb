Spark Structured Streaming Triggers
What is a Trigger?
A trigger decides when and how often Spark processes streaming data.
Simple:
Trigger = WHEN the streaming query processes data.

1. Default Trigger
If we don't specify any trigger, Spark runs micro-batches as quickly as possible. It is equivalent to processingTime='0 seconds'. 
Syntax
query = (
df.writeStream
.format("console")
.start()
)
How it works
New data
↓
Batch 1
↓
Batch 2
↓
Batch 3
↓
Continues...
Spark starts the next micro-batch as soon as it can.
Use case
•	Near-real-time processing
•	Frequently updating dashboards
•	Streaming workloads where low latency is required
Interview answer
Default trigger processes data in micro-batches as quickly as possible when no trigger interval is specified.
Remember: Default = ASAP
2. Processing Time Trigger
Also called a fixed interval micro-batch trigger.
Spark attempts to start micro-batches at the interval you specify
Syntax
Python
query = (
df.writeStream
.trigger(processingTime="10 seconds")
.format("console")
.start()
)
Example
10:00:00 → Batch
10:00:10 → Batch
10:00:20 → Batch
10:00:30 → Batch
Important clarification
Suppose you configured:
processingTime="10 seconds"
but one batch takes 15 seconds.
Spark doesn't start another overlapping batch at the 10-second point. Once the previous batch finishes, the next one can start immediately if it is behind schedule. 
Use case
•	Process data every 10 seconds
•	Periodic aggregations
•	Balance latency and resource usage
Interview answer
Processing Time trigger runs micro-batches at a specified time interval.
Remember: Processing Time = Every X seconds/minutes
3. Available Now Trigger ⭐
This is an important one for Databricks.
It processes all data that is available when the query starts, potentially using multiple micro-batches, and then stops. 
Syntax
query = (
df.writeStream
.trigger(availableNow=True)
.format("console")
.start()
)
Example
Suppose 1,000 files are waiting:
1000 files available
↓
Batch 1
↓
Batch 2
↓
Batch 3
↓
All available data processed
↓
STOP
It doesn't necessarily process everything in one batch.
Use case
Very useful for:
•	Scheduled incremental ETL
•	Hourly jobs
•	Daily jobs
•	Auto Loader ingestion
•	Batch-like processing using Structured Streaming
Interview answer
Available Now processes all currently available data using one or more micro-batches and then stops.
Remember: Available Now = Process everything available → STOP
Azure Databricks specifically recommends this approach for incremental batch processing. 
4. Once Trigger
Syntax
query = (
df.writeStream
.trigger(once=True)
.format("console")
.start()
)
It processes one micro-batch and then terminates. [spark.apache.org]
Important difference
Once:
ONE micro-batch
↓
STOP
Available Now:
Plain Text
All currently available data
↓
Batch 1 → Batch 2 → Batch 3
↓
STOP
Important for Databricks interviews
Trigger.Once is deprecated in Databricks Runtime 11.3 LTS and later. Databricks recommends AvailableNow instead. 
Interview answer
Once trigger processes one micro-batch and stops. In modern Databricks workloads, Available Now is preferred.
Remember: Once = One batch → STOP
5. Continuous Trigger
Apache Spark provides an experimental Continuous Processing trigger designed for very low latency. 
Spark OSS syntax
Python
df.writeStream \
.trigger(continuous="1 second") \
.format("console") \
.start()
Concept
Instead of standard micro-batch processing:
Batch 1 → wait → Batch 2 → wait → Batch 3
Continuous processing is designed more like:
Record → Process
Record → Process
Record → Process
...
Important for your Databricks interview
Azure Databricks currently says the Spark Continuous Processing trigger is not supported. 
So don't say:
"We normally use continuous trigger in Databricks."
Better interview answer:
Continuous Processing is an experimental Apache Spark feature for low-latency processing, but it isn't supported as a trigger in Azure Databricks.
________________________________________
6. Real-Time Mode
This is a newer point worth knowing for current Databricks interviews.
Azure Databricks provides a real-time mode for ultra-low-latency operational workloads, such as fraud detection and real-time personalization. 
Example:
Python
df.writeStream \
.trigger(realTime="5 minutes") \
.start()
The 5 minutes here represents the duration of the long-running batch/checkpoint interval, not a 5-minute latency for every record. Records are processed as they arrive during the batch. 
For basic interviews, know this exists. For most normal Databricks data-engineering questions, focus mainly on Default, Processing Time, and Available Now.
