unionByName() lets you merge/union two DataFrames by matching columns based on their names. It can also handle DataFrames with different schemas when allowMissingColumns=True, filling missing columns with null values.

```python
data1 = [(1,'trinadh','male')]
schema1=['id','name' ,'gender']
data2 = [(1,'srinadh','2000')]
schema2=['id','name' ,'salary']

df1=spark.createDataFrame(data1,schema1)
df2=spark.createDataFrame(data2,schema2)

df1.union(df2).display()
df1.unionByName(df2, allowMissingColumns=True).display()
```
