**Restore Command**

In Databricks, the RESTORE command is used to bring back a table to a previous version. Think of it like a "time machine" for your data.
 **Why would you use it?**

Imagine you made a mistake—like accidentally deleting or updating important data in a table. Instead of manually fixing it, you can use RESTORE to go back to a version of the table before the mistake happened.

**🧭 What Does RESTORE Do?**

The RESTORE command lets you revert a Delta table to a previous version. This is like pressing an "undo" button for your data.
1️⃣ You accidentally deleted or modified data
- Scenario: You ran a query that deleted thousands of rows from your orders table by mistake.
- Problem: The data is gone, and manually recovering it would be hard.
- Solution: Use RESTORE to go back to the version before the deletion.
- RESTORE TABLE customer_data TO VERSION AS OF 8;


### create table

```python
from delta.tables import *

DeltaTable.create(spark) \
    .tableName("tri.tri.scd2Demo1") \
    .addColumn("pk1", "INT") \
    .addColumn("pk2", "STRING") \
    .addColumn("dim1", "INT") \
    .addColumn("dim2", "INT") \
    .addColumn("dim3", "INT") \
    .addColumn("dim4", "INT") \
    .addColumn("active_status", "STRING") \
    .addColumn("start_date", "TIMESTAMP") \
    .addColumn("end_date", "TIMESTAMP") \
    .execute()
```

### # Create DeltaTable Instance Using Table Name

```python
from delta.tables import *

targetTable = DeltaTable.forName(
    spark,
    "tri.tri.scd2Demo1"
)
```

```python
display(targetTable.history())
```

### Restore Delta Table Using Timestamp

```python
targetTable.restoreToTimestamp(
    '2026-10-25T10:58:37.000+00:00'
)
```

```python
%sql
select * from tri.tri.scd2demo1
```

### Restore Delta Table Using Version Number


```python
targetTable.restoreToVersion(2)
```

```python
%sql
select * from tri.tri.scd2demo1
```
