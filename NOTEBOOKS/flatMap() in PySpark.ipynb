**✅ What is flatMap() in PySpark?**
- •	flatMap() is an RDD transformation.
- •	It applies a function to each element of the RDD and then flattens the result into a single RDD.
- •	Unlike map(), which returns one element per input, flatMap() can return zero, one, or many elements per input.
- •	Not available on DataFrames. For DataFrames, use explode() to flatten arrays.

**✅ Key Difference: map() vs flatMap()**
- •	map() → 1 input → 1 output (or tuple)
- •	flatMap() → 1 input → multiple outputs, and flattens them into a single RDD.

**✅ When to Use flatMap()?**
- •	When you need to split or expand elements into multiple parts (e.g., splitting sentences into words).
- •	When you want a flattened structure instead of nested lists.


```python
# Sample data
data = ['maheer shaik', 'wafa shaik']
# Create RDD
rdd = spark.sparkContext.parallelize(data)
print("Original RDD:")
for item in rdd.collect():
    print(item)
# Apply flatMap to split each string into words
rdd1 = rdd.flatMap(lambda x: x.split(' '))
print("After flatMap:")
for item in rdd1.collect():
    print(item)

```

Output:
Original RDD:
maheer shaik
wafa shaik

After flatMap:
maheer
shaik
wafa
shaik

**RDD Function	DataFrame Replacement**
- map()	select(), withColumn()
- flatMap()	explode()
- filter()	filter(), where()
- reduceByKey()	groupBy().agg()
- mapPartitions()	foreachPartition() / Pandas UDFs (when needed)

**Simple Way to Remember**
- map() → modify a row → withColumn() / select()
- flatMap() → split one row into many rows → explode()
