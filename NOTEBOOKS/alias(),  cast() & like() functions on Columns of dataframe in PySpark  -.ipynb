# 🔹 alias() in PySpark
# Definition:
# The alias() function is used to rename a column in a DataFrame. This is especially useful when:
# •	You want more readable or meaningful column names.
# •	You're preparing data for output or reporting.
# •	You're avoiding column name conflicts during joins or aggregations

```python

# Example:
data = [(1, 'maheer', 'm', 2000), (2, 'tri', 'm', 5000), (3, 'heer', 'f', 2600)]
schema = ['id', 'name', 'gender', 'salary']
df = spark.createDataFrame(data, schema)
df.select(df.id.alias('emp_id'), df.name.alias('emp_name')).show()
```

# 🔹  cast() – Convert the Data Type
# ✅ Definition:
# The cast() function is used to convert the data type of a column in a DataFrame.

```python

# ✅ Example:
data = [(1, 'maheer', 'm', 2000), (2, 'tri', 'm', 5000), (3, 'heer', 'f', 2600)]
schema = ['id', 'name', 'gender', 'salary']
df = spark.createDataFrame(data, schema)
df1 = df.select(df.salary.cast('int'))
df1.printSchema()
```

# 🔹  like() – Pattern Matching
# ✅ Definition:
# The like() function is used to filter rows based on pattern matching, similar to SQL's LIKE clause.

```python
# ✅ Example:
data = [(1, 'maheer', 'm', 2000), (2, 'tri', 'm', 5000), (3, 'heer', 'f', 2600)]
schema = ['id', 'name', 'gender', 'salary']
df = spark.createDataFrame(data, schema)
df.filter(df.name.like('m%')).show()
```
