**🔹 Definition of agg() in PySpark**

The agg() function in PySpark is used in combination with groupBy() to apply multiple aggregate functions simultaneously on grouped data. It provides flexibility to compute various statistics like count, sum, min, max, avg, etc., all in one go.


```python
data = [
    (1, 'maheer', 'M', 5000, 'IT'),
    (2, 'wafa', 'M', 6000, 'IT'),
    (3, 'Asif', 'F', 2500, 'Payroll'),
    (4, 'sarfaraj', 'M', 4000, 'HR'),
    (5, 'pyarijaan', 'F', 2000, 'HR'),
    (6, 'Mahaboob', 'M', 2000, 'Payroll'),
    (7, 'ayesha', 'F', 3000, 'IT')
]
schema = ['id', 'name', 'gender', 'salary', 'dep']
df = spark.createDataFrame(data, schema)
from pyspark.sql.functions import count, min, max
df.groupBy('dep').agg(
    count('*').alias('countOfEmps'),
    min('salary').alias('minSal'),
    max('salary').alias('maxSal')
).show()
```
