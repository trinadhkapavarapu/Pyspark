- **Short Note:** Distinct vs DropDuplicates in PySpark
- Both methods are used to remove duplicates from a DataFrame, but they differ in scope and flexibility.
- **distinct()**
- •	Removes fully duplicate rows across all columns.
- •	Equivalent to SELECT DISTINCT * in SQL.
- •	No subset elimination: You cannot specify which columns to consider.
- •	Example: df.distinct()
- **dropDuplicates()**
- •	Removes duplicates based on specific columns.
- •	More flexible when you want to keep unique rows based on a subset of columns.
- •	Example: df.dropDuplicates(["column1", "column2"])

```python
simpleData = [
    (111, "James", 8),
    (222, "Mike", 5),
    (111, "James", 8),
    (222, "Mike", 12)
]
columns = ["id", "name", "score"]
df = spark.createDataFrame(data=simpleData, schema=columns)
df.display()


df.distinct().show()#-------------remove duplicates
df.dropDuplicates(['id']).show()#--------------------- remove duplicates column based
```
