write dataframe into json file using pyspark        



```python
data = [(1,'jhansi'),(2,'jhanu')]
schema = ['id','name']
df = spark.createDataFrame(data = data,schema = schema)

df.write.json(path='/Volumes/tri/tri/tri_files/OUTPUT/',mode = 'overwrite')
display(spark.read.json(path='/Volumes/tri/tri/tri_files/OUTPUT/'))
```

we can use mode = ignore /overwrite/append/error

```python

df.write.json(path='dbfs:/tmp/emps',header=true,mode = 'error')#-data there it gives
df.write.json(path='dbfs:/tmp/emps',header=true,mode = 'overwrite')#-overwrite existing data
df.write.json(path='dbfs:/tmp/emps',header=true,mode = 'ignore')#-throw error data there
df.write.json(path='dbfs:/tmp/emps',header=true,mode = 'append')#-it will add data

```

note: for json no need to use header = true
