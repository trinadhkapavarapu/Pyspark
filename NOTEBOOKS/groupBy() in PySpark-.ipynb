**🔹 PySpark groupBy() – Complete Definition**

The groupBy() function in PySpark is used to group rows in a DataFrame that share the same values in one or more columns. It is conceptually similar to the SQL GROUP BY clause and is essential for performing aggregate operations on grouped data.



**🔹 Supported Aggregate Functions**
- **Function	Description**
- count()	Returns the number of rows in each group
- sum()	Returns the total of values in each group
- avg() / mean()	Returns the average of values in each group
- min()	Returns the minimum value in each group
- max()	Returns the maximum value in each group
- agg()	Allows multiple aggregate functions to be applied at once


```python

#Create dataframe

data = [
    (1, 'maheer', 'M', 5000, 'IT'),
    (2, 'wafa', 'M', 6000, 'IT'),
    (3, 'asif', 'F', 2500, 'Payroll'),
    (4, 'sarfaraj', 'M', 4000, 'HR'),
    (5, 'pyarjaan', 'F', 2000, 'HR'),
    (6, 'Mahaboob', 'M', 2000, 'Payroll'),
    (7, 'ayesha', 'F', 3000, 'IT')
]
schema = ['id', 'name', 'gender', 'salary', 'dep']
df = spark.createDataFrame(data, schema)
```

```python
#🔹 Supported Aggregate Functions

#Count by Department
df.groupBy('dep').count().show()
#Minimum Salary by Department
df.groupBy('dep').min('salary').show()
#Maximum Salary by Department
df.groupBy('dep').max('salary').show()
#Average Salary by Department
df.groupBy('dep').avg('salary').show()
#Sum of Salary by Department
df.groupBy('dep').sum('salary').show()
#Mean (Average) of Salary by Department
df.groupBy('dep').mean('salary').show()
```
