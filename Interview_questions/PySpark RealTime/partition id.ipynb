**Find the Skewness in the data by querying the number of records per PARTITION. Also, RANK them accordingly.**

Which partition has more data 

```python
from pyspark.sql.functions import *
```

```python
df = spark.read.format("csv")\
        .option("header", True)\
        .option("inferSchema", True)\
        .load("/FileStore/rawsource/sales_data_first.csv")

df.display()
```

```python
df = df.withColumn("partition_id", spark_partition_id())

df.display()
```

We have only 1 partioon,so better to repartiotion

```python
df =df.repartition(4)
```

**In reall world scenario we have lots partitions unable to see full data to count how many partitions there –so do group by and count **

```python
df = df.withColumn("partition_id", spark_partition_id()).groupBy("partition_id").count()

df.display()
```
