**Column**
- PySpark Column class represents a single Column in a DataFrame.
- pyspark.sql.Column class provides several functions to work with DataFrame to manipulate the Column values, evaluate the boolean expression to filter rows, retrieve a value or part of a value from a DataFrame column.
- One of the simplest ways to create a Column class object is by using PySpark lit() SQL function.

#✅ 1) lit() and col basics

```python
#✅ 1) lit() and col basics

from pyspark.sql import SparkSession
from pyspark.sql.functions import lit, col

spark = SparkSession.builder.getOrCreate()

# lit() creates a Column expression (not a Python string)
col1 = lit('abcd')
print(type(col1)) # <class 'pyspark.sql.column.Column'>

# Use lit() inside a DataFrame context
spark.range(1).select(col1.alias('const')).show()




```

# ✅ 2) Create a DataFrame, add a new column, select columns

```python
#✅ 2) Create a DataFrame, add a new column, select columns

from pyspark.sql.functions import lit, col

# Correct: strings must be quoted; schema is a list/tuple of column names
data = [(1, 'tri', 2000), (2, 'trnnni', 2000), (3, 'tri', 2000)]
columns = ('id', 'name', 'salary')

df = spark.createDataFrame(data=data, schema=columns)

# Add a new column using lit()
df1 = df.withColumn('country', lit('usa')) # <-- "new column (col)"
df1.show()
df1.printSchema()

# Access / select a column in three ways
df1.select('country').show()
df1.select(df1['country']).show()
df1.select(col('country')).show()

# If you want to select 'gender', it must exist; add it first:
df2 = df1.withColumn('gender', lit('male'))
```

# #✅ 3) Struct (nested) column — define schema and access nested fields

```python
#✅ 3) Struct (nested) column — define schema and access nested fields

from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import col

# Define the inner struct (props)
propsType = StructType([
StructField('hair', StringType(), True),
StructField('eye', StringType(), True),
])

# Define the full schema
schema = StructType([
StructField('id', IntegerType(), False),
StructField('name', StringType(), True),
StructField('props', propsType, True),
StructField('salary', IntegerType(), True),
])

# Data must match the schema order; tuples for the struct field
data_struct = [
(1, 'tri', ('black', 'brown'), 3000),
(2, 'sri', ('black', 'blue'), 4000),
]

df3 = spark.createDataFrame(data_struct, schema=schema)
df3.show(truncate=False)
df3.printSchema()

# Access nested (struct) fields — 3 ways
df3.select('props.hair').show() # by string path
df3.select(df3.props.hair).show() # via attribute chaining
df3.select(col('props.hair')).show() # via col()
```
