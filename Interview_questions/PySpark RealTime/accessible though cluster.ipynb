**Query the DATAFRAME using SQL as a TABLE. Also, this object should be accessible throughout the CLUSTER.**

```python
from pyspark.sql.functions import *
```

```python
df = spark.read.format("csv")\
        .option("header",True)\
        .option("inferSchema",True)\
        .load("/FileStore/rawsource/sales_data_first.csv")

df.display()
```

```python
df.createOrReplaceTempView("temp_view")
```

```python
%sql
SELECT * FROM temp_view
```

**But it will work on same notebook not othere notebook


Creating global temp view**


```python
df.createOrReplaceGlobalTempView("global_view")
```

```python
%sql
SELECT * FROM global_temp.global_view
```

**Now works select * from in any notebook with same cluster**
