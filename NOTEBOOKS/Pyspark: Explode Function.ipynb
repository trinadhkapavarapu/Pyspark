**What is Explode**

It is a PySpark function that creates a new row for each element in an array or map. By default, it uses the column name col for array elements and key and value for map elements.

**Explode –**When an array is passed to this function, it creates a new row for each element in array. When a map is passed, it creates two new columns one for key and one for value and each element in map split into the rows. if the array or map is null, that row is eliminated.

**Explode_outer –** Unlike explode, if the array or map is null, explode_outer returns null.

**Posexplode –**When array or map is passed, creates positional column also for each element. Ignores the null elements.

**Posexplode_outer –** Unlike posexplode, if the array or map is null, explode_outer returns null.

### creates a DataFrame with an array column

```python
array_appliance = [
    ('Raja', ['TV', 'Refrigerator', 'Oven', 'AC']),
    ('Raghav', ['AC', 'Washing machine', None]),
    ('Ram', ['Grinder', 'TV']),
    ('Ramesh', ['Refrigerator', 'TV', None]),
    ('Rajesh', None)
]
df_app = spark.createDataFrame(data=array_appliance, schema=['name', 'Appliances'])
df_app.printSchema()
display(df_app)
```

```python
#Method 1
from pyspark.sql.functions import explode, col
df_app.show()
df1=df_app.withColumn('appliances1',explode(col('appliances')))
df1.show()
```

```python
#method2
from pyspark.sql.functions import explode,col
df_app.show()
df1=df_app.withColumn('appliances1',explode(col('appliances'))).select('name','appliances1')
df1.show()
```

```python
#method 3
from pyspark.sql.functions import explode
df2 = df_app.select(df_app.name, explode(df_app.Appliances))
df_app.printSchema()
display(df_app)
df2.printSchema()
display(df2)
```

### create a DataFrame with a map column

```python
from pyspark.sql.types import *

schema = StructType([
    StructField("name", StringType(), True),
    StructField("Brand", MapType(StringType(), StringType()), True)
])

df_brand = spark.createDataFrame(map_brand, schema=schema)
```

```python
map_brand = [
    ('Raja', {'TV': 'LG', 'Refrigerator': 'Samsung', 'Oven': 'Philipps', 'AC': 'Voltas'}),
    ('Raghav', {'AC': 'Samsung', 'Washing machine': 'LG'}),
    ('Ram', {'Grinder': 'Preethi', 'TV': ''}),
    ('Ramesh', {'Refrigerator': 'LG', 'TV': 'Croma'}),
    ('Rajesh', None)
]
df_brand = spark.createDataFrame(data=map_brand, schema=schema)
df_brand.printSchema()
display(df_brand)

from pyspark.sql.functions import explode
df3 = df_brand.select(df_brand.name, explode(df_brand.Brand))
df3.printSchema()
display(df3)
```

### Explode outer to consider null values

```python
from pyspark.sql.functions import explode_outer
# Exploding array column with nulls included
display(df_app.select(df_app.name, explode_outer(df_app.Appliances)))
# Exploding map column with nulls included
display(df_brand.select(df_brand.name, explode_outer(df_brand.Brand)))
```

### Positional Explode

```python

from pyspark.sql.functions import posexplode
# Explode array column with position
display(df_app.select(df_app.name, posexplode(df_app.Appliances)))
# Explode map column with position
display(df_brand.select(df_brand.name, posexplode(df_brand.Brand)))
```

### Positional Explode with null

```python

from pyspark.sql.functions import posexplode_outer
# Explode array column with position, including nulls
display(df_app.select(df_app.name, posexplode_outer(df_app.Appliances)))
# Explode map column with position, including nulls
display(df_brand.select(df_brand.name, posexplode_outer(df_brand.Brand)))
```
