**🔷 What is a Delta Table?**
- A Delta table is a special kind of table in Databricks that uses the Delta Lake format. It’s like a regular table, but with extra powers:
- •	It keeps track of changes (like a version history).
- •	It supports ACID transactions (safe and reliable updates).
- •	It allows fast reads and writes.
- •	You can time travel to see old versions of the data.
________________________________________
**✅ Delta Table Instance 
What it is:**
- •	A Delta Table Instance is a programmatic reference (object) to an existing Delta table.
- •	It can be thought of as a handle to the underlying Delta table, allowing you to interact with and manage the table through the Delta Lake API (PySpark/Scala).
- •	The instance itself does not store data; it simply provides access to the Delta table and its operations.

**Usage:**
- You can use a Delta Table Instance to perform Delta Lake operations such as:
- •	UPDATE, DELETE, MERGE, VACUUM, DESCRIBE, HISTORY
- •	These operations are typically performed using the Delta Lake API (PySpark/Scala) in Databricks. Equivalent operations can also be executed using SQL statements.



- **✅ Delta Table Instance Creation Syntax**

**Using Path**
- DeltaTable.forPath(spark, "/path/to/table")

**Using Table Name**
- DeltaTable.forName(spark, "table_name")


# Create Delta Table

```python
from delta.tables import *

DeltaTable.createOrReplace(spark) \
    .tableName("tri.tri.employee_demo") \
    .addColumn("emp_id", "INT") \
    .addColumn("emp_name", "STRING") \
    .addColumn("gender", "STRING") \
    .addColumn("salary", "INT") \
    .addColumn("Dept", "STRING") \
    .property("description", "table created for demo purpose") \
    .execute()
```

```python
%sql

INSERT INTO tri.tri.employee_demo
VALUES (100, "Stephen", "M", 2000, "IT");

INSERT INTO tri.tri.employee_demo
VALUES (200, "Philipp", "M", 8000, "HR");

INSERT INTO tri.tri.employee_demo
VALUES (300, "Lara", "F", 6000, "SALES");
```

```python
%sql
SELECT * FROM tri.tri.employee_demo;
```

# Create Delta Table Instance Using Path

```python
deltaInstance1 = DeltaTable.forPath(
    spark,
    "/FileStore/tables/delta/createtable"
)
```

```python
display(deltaInstance1.toDF())
```

# Method 1: Delete Using SQL

```python
%sql
DELETE FROM tri.tri.employee_demo
WHERE emp_id = 100;
```

```python
%sql
SELECT * tri.tri.FROM employee_demo;
```

# Method 2: Delete Using SQL Predicate (Table Instance)

```python
deltaInstance1.delete("emp_id = 200")
```

```python
display(deltaInstance1.toDF())
```

# Create Delta Table Instance Using Table Name

```python
from delta.tables import *

deltaInstance2 = DeltaTable.forName(
    spark,
    "tri.tri.employee_demo"
)
```

```python
display(deltaInstance2.toDF())
```

```python
%sql
describe history tri.tri.employee_demo
```

```python
display(deltaInstance2.history())
```
