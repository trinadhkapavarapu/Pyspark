**array()**
Use array() function to create a new array column by merging the data from multiple columns.


```python

# Create sample data
data = [(1, 'trinadh', 'dotnet', 'azure'),
        (2, 'srinadh', 'java', 'aws')]
# Define schema
schema = ['id', 'name', 'primaryskills', 'secondaryskills']
# Create DataFrame
df = spark.createDataFrame(data, schema)
display(df)
# Print schema
df.printSchema()
# Import necessary functions
from pyspark.sql.functions import split, col, array
# Show original DataFrame
df.show()
# Add a new column 'skillsArray' combining primary and secondary skills
df1 = df.withColumn('skillsArray', array(col('primaryskills'), col('secondaryskills')))
# Show updated DataFrame
df1.show()

```


**array_intersect()** PySpark function is used to return common elements present in two arrays without any duplication.

```python
#Create a sample dataframe
empDF = [
    ('John', [4, 6, 7, 9, 2], [1, 2, 3, 7, 7]),
    ('David', [7, 5, 1, 4, 7, 1], [3, 2, 8, 9, 4, 9]),
    ('Mike', [3, 9, 1, 6, 2], [1, 2, 3, 5, 8])
]
df = spark.createDataFrame(data=empDF, schema=['Name', 'Array_1', 'Array_2'])
df.show()

#apply array_intersect
from pyspark.sql import functions as F
# Create a new column 'Intersect' with common elements from Array_1 and Array_2
df_intersect = df.withColumn("Intersect", F.array_intersect(df["Array_1"], df["Array_2"]))
df_intersect.show()
```


**array_except()** PySpark function is used to return the list of elements which are present in first array but not in second array.

```python
#Create sample dataframe
empDF = [
    ('John', [4, 6, 7, 9, 2], [1, 2, 3, 7, 7]),
    ('David', [7, 5, 1, 4, 7], [3, 2, 8, 9, 4, 9]),
    ('Mike', [3, 9, 1, 6, 2], [1, 2, 3, 5, 8])
]
df = spark.createDataFrame(data=empDF, schema=['Name', 'Array_1', 'Array_2'])
# Show DataFrame
df.show()
#apply array_except
from pyspark.sql import functions as F
# Create a new column 'Intersect' with elements in Array_1 not in Array_2
df_except = df.withColumn("Intersect", F.array_except(df["Array_1"], df["Array_2"]))
df_except.show()
```


**array_sort()** PySpark function is used to sort the elements within the array in ascending order.

```python
#Create data frame
empDF = [
    ('John', [4, 6, 7, 9, 2]),
    ('David', [7, 5, 1, 4, 7, 1]),
    ('Mike', [3, 9, 1, 6, 2])
]
df = spark.createDataFrame(data=empDF, schema=['Name', 'Array_col'])
df.show()

#apply sort on array
from pyspark.sql import functions as F
# Create a new column 'Sorted' by sorting the values in 'Array_col'
df_sort = df.withColumn("Sorted", F.array_sort(df["Array_col"]))
df_sort.show()
```


**arrays_zip()** PySpark function is used to return a merged array of structs in which the N-th struct contains all N-th values of input arrays.

```python
#Create sample dataframe
array_data = [
    ("John", 4, 1),
    ("John", 6, 2),
    ("David", 7, 3),
    ("Mike", 3, 4),
    ("David", 3, 4),
    ("John", 7, 5),
    ("David", 1, 8),
    ("John", 9, 9),
    ("David", 7, 4),
    ("Mike", 5, 2),
    ("Mike", 5, 8),
    ("John", 2, 7),
    ("David", 1, 9)
]
array_schema = ["Name", "Score_1", "Score_2"]
arrayDF = spark.createDataFrame(data=array_data, schema=array_schema)
display(arrayDF)



#Convert sample dataframe into array dataframe
from pyspark.sql import functions as F
masterDF = arrayDF.groupby("Name").agg(
    F.collect_list('Score_1').alias('Array_Score_1'),
    F.collect_list('Score_2').alias('Array_Score_2')
)
display(masterDF)
masterDF.printSchema()
```

```python
#Apply arrays_zip function on array df
from pyspark.sql import functions as F
arr_zip_df = masterDF.withColumn(
    "Zipped_value",
    F.arrays_zip("Array_Score_1", "Array_Score_2")
)
arr_zip_df.show(10, False)
```

# Practical use case to flatten data using arrays_zip and explode

```python
#Create Sample Nested DataFrame

empDF = [
    ('Sales_dept', [
        {'emp_name': 'John', 'salary': '1000', 'yrs_of_service': '10', 'Age': '33'},
        {'emp_name': 'David', 'salary': '2000', 'yrs_of_service': '15', 'Age': '40'},
        {'emp_name': 'Nancy', 'salary': '3000', 'yrs_of_service': '20', 'Age': '45'},
        {'emp_name': 'Mike', 'salary': '4000', 'yrs_of_service': '8', 'Age': '38'},
        {'emp_name': 'Rosy', 'salary': '6000', 'yrs_of_service': '6', 'Age': '31'}
    ]),
    ('HR_dept', [
        {'emp_name': 'Edwin', 'salary': '5000', 'yrs_of_service': '10', 'Age': '33'},
        {'emp_name': 'Tomas', 'salary': '7000', 'yrs_of_service': '4', 'Age': '26'},
        {'emp_name': 'Sarah', 'salary': '12000', 'yrs_of_service': '22', 'Age': '49'},
        {'emp_name': 'Stella', 'salary': '15000', 'yrs_of_service': '25', 'Age': '52'},
        {'emp_name': 'Kevin', 'salary': '40000', 'yrs_of_service': '5', 'Age': '27'}
    ])
]
df_brand = spark.createDataFrame(data=empDF, schema=["Department", "Employee"])
df_brand.printSchema()
display(df_brand)


```

```python
#apply array_zip
from pyspark.sql import functions as F
df_brandZip = df_brand.withColumn("Zip", F.arrays_zip(df_brand["Employee"]))
display(df_brandZip)
```

```python
#Apply explode
df_brand_exp = df_brandZip.withColumn("Explode", F.explode(df_brandZip.Zip))
display(df_brand_exp)
```

# Code to Flatten Exploded Struct

```python
df_brand_output = df_brand_exp.withColumn("employee_emp_name", df_brand_exp['Explode.Employee.emp_name']) \
    .withColumn("employee_yrs_of_service", df_brand_exp['Explode.Employee.yrs_of_service']) \
    .withColumn("employee_salary", df_brand_exp['Explode.Employee.salary']) \
    .withColumn("employee_Age", df_brand_exp['Explode.Employee.Age']) \
    .drop("Explode").drop("Zip").drop("Employee")
display(df_brand_output)
```
