multiLine = True: Spark correctly reads JSON records that span multiple lines.

```python
df = spark.read.format("json") \
    .option("inferSchema", True) \
    .option("multiLine", True) \
    .load("/Volumes/pyspark_cata/source/db_volume/jsonData/")

display(df)
```

```python
df.schema
```

```python
from pyspark.sql.functions import *

df_cust = df.select(
    "customer.customer_id",
    "customer.email",
    "customer.location.city",
    "customer.location.country"
)

display(df_cust)
```

```python
from pyspark.sql.functions import *

df_cust = df.select(
    "customer.customer_id",
    "customer.email",
    "customer.location.city",
    "customer.location.country",
    "*"
).drop("customer")

df_cus*_upd = (
    df_cust.withColumn("delivery_updates", explode("delivery_updates"))
           .withColumn("items", explode("items"))
           .select("*")
)

display(df_cust_upd)
```
