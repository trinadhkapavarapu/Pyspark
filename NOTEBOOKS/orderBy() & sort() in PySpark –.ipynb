**orderBy() & sort()**

- You can use either orderBy() or sort() function of a PySpark DataFrame to arrange rows in ascending or descending order based on one or more columns.
- By default, sorting is performed in ascending order. You can explicitly specify the sort order using asc() and desc() functions.

```python
data = [
    (1, 'tri', 'm', 2000, 'it'),
    (2, 'tri', 'm', 4000, 'hr'),
    (3, 'sri', 'f', 2000, 'payroll'),
    (4, 'tri', 'm', 2000, 'hr')
]
schema = ['id', 'name', 'gender', 'salary', 'dep']
df = spark.createDataFrame(data, schema)
# Sort by department and salary (ascending)
df.sort('dep', 'salary').show()
# Sort by department and salary (descending)
df.sort(df.dep.desc(), df.salary.desc()).show()
# Order by department and salary (ascending)
df.orderBy('dep', 'salary').show()
# Order by department and salary (descending)
df.orderBy(df.dep.desc(), df.salary.desc()).show()





```

**🔹 2. asc() and desc() in PySpark**
**Definition:**
- •	asc() sorts a column in ascending order.
- •	desc() sorts a column in descending order.
- These are used with sort() or orderBy() to control the order of rows in the DataFrame.

```python

#Example:
#🔹  Sorting with asc() and desc()
#Code:
data = [(1, 'maheer', 'm', 2000), (2, 'tri', 'm', 5000), (3, 'heer', 'f', 2600)]
schema = ['id', 'name', 'gender', 'salary']
df = spark.createDataFrame(data, schema)
df.sort(df.name.desc()).show()
```
