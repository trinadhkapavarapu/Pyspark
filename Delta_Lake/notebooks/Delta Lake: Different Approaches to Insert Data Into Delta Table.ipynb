# 2.DataFrame Insert Example in Databricks

- Sure! Here's a simple explanation of the three output modes used in Structured Streaming in Databricks (and Spark):
- ________________________________________
- 🔁 Append Mode
- •	What it does: Only new rows that arrive in the stream are written to the output.
- •	Example: Imagine you're getting new orders in a shop. Append mode will only write the new orders to the database or file, not the old ones.
- ________________________________________
- 📋 Complete Mode
- •	What it does: Writes the entire result table every time there's an update.
- •	Example: If you're tracking total sales per product, every time a new sale happens, it will rewrite the full sales summary for all products.
- ________________________________________
- 🔄 Update Mode
- •	What it does: Writes only the rows that changed since the last update.
- •	Example: If only the sales for "Product A" changed, then only that row will be updated in the output—not the whole table.
- ________________________________________
- Summary Table
- Mode	What it writes	Good for...
- Append	Only new rows	Logs, events, new transactions
- Complete	Entire result table	Aggregates, summaries
- Update	Only changed rows	Efficient updates to dashboards




```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

employee_data = [(200, "Philipp", "M", 8000, "HR")]

employee_schema = StructType([
    StructField("emp_id", IntegerType(), False),
    StructField("emp_name", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("dept", StringType(), True)
])

df = spark.createDataFrame(data=employee_data, schema=employee_schema)

display(df)
```

```python
#Append DataFrame to Existing Table

df.write.format("delta") \
    .mode("append") \
    .saveAsTable("tri.tri.employee_demo")

display(spark.sql("SELECT * FROM tri.tri.employee_demo"))
```

# 3.DataFrame insertInto() Method Example in Databricks

- 1. df1.write.insertInto('employee_demo', overwrite=False)
- Simple explanation:
- This takes the data from df1 and adds it to the employee_demo table.
- •	Existing data stays as it is.
- •	New records are added to the table.
- •	Nothing gets deleted.
- 🧠 Think of it like: Adding new rows to an existing Excel sheet without removing the old rows.
- Remember: overwrite=False = Append/Add data
________________________________________
- 2. df1.write.insertInto('employee_demo', overwrite=True)
- Simple explanation:
- This takes the data from df1 and replaces the existing data in employee_demo.
- •	Existing data is removed.
- •	Only the new data from df1 remains.
- 🧠 Think of it like: Clearing an Excel sheet and then loading fresh data into it.
- Remember: overwrite=True = Replace existing data
________________________________________
- 📌 Shortcut:
- •	overwrite=False → Add new data
- •	overwrite=True → Replace old data with new data
- 

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

employee_data = [(300, "Lara", "F", 6000, "SALES")]

employee_schema = StructType([
    StructField("emp_id", IntegerType(), False),
    StructField("emp_name", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("Dept", StringType(), True)
])

df1 = spark.createDataFrame(data=employee_data, schema=employee_schema)

display(df1)
```

```python
#Insert Data into Existing Table Using insertInto()
df1.write.insertInto("tri.tri.employee_demo", overwrite=False)
display(spark.sql("SELECT * FROM tri.tri.employee_demo"))
```
