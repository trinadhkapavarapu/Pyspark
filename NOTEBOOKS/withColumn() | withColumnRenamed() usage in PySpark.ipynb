Main focus :-
- Change the data type of an existing column
-  Add a new column with a constant (literal) value
- Create a new column based on existing column(s) using calculations or transformations
- Update the values of an existing column

```python
#Data set

employee_data = [
    (10, "Raj", "Kumar", "1999", 100, "M", 2000),
    (20, "Rahul", "Rajan", "2002", 200, "F", 8000),
    (30, "Raghav", "Manish", "2010", 100, None, 6000),
    (40, "Raja", "Singh", "2004", 100, "F", 7000),
    (50, "Rama", "Krish", "2008", 400, "M", 1000),
    (60, "Ravi", "Kutty", "2012", 400, None, 5000),
    (70, "Kumar", "Chand", "2004", 600, "M", 5000)
]
employee_schema = [
    "employee_id",
    "first_name",
    "last_name",
    "doj",
    "employee_dept_id",
    "gender",
    "salary"
]
df = spark.createDataFrame(data=employee_data, schema=employee_schema)
display(df)
```

```python
#Convert data type
from pyspark.sql.functions import col
df1=df.withColumn(colName='salary',col=col('salary').cast('integer'))
df1.show()
df1.printSchema()
```

```python
#Add a column using constant literal.
from pyspark.sql.functions import lit
df2 = df1.withColumn('country', lit('usa'))  #---------add column, value
df2.show()
```

```python
#Create a new column based on existing column(s) using calculations or transformations

from pyspark.sql.functions import concat, lit,col


df3 = df2.withColumn("Bonus", df2.salary * 0.1) \
    .withColumn("Name", concat("first_name", lit(" "), "last_name"))
display(df3)
```

```python

#Update the values of an existing column
from pyspark.sql.functions import concat, lit,col
df4=df3.withColumn('salary',col('salary')*2)
df4.show()
```

# withColumnRenamed() usage in PySpark

```python
#Changing column name-

empDF_RenameColumn = df4.withColumnRenamed("Name", "Full_Name")
display(empDF_RenameColumn)
```

```python
#drop column
empDF_DropColumn = empDF_RenameColumn.drop("name")
empDF_DropColumn.show()
```
