```python
data = [(1,'tri12321423fjkadbv zfvnzdjvdjfvnDnvdnfvd'),
 (2,'nadhsdnfksdngvadfgvnmdfnbnSNBDKnb'),
(3,'nadhsdnfksdngvadfgvnmdfnbnSNBDKnb')]
schema = ['id','comments']
df = spark.createDataFrame(data=data,schema=schema)


df.show(n=3,truncate=False,vertical=True)#-----------mention number of rows,show full data,vertical
df.show(truncate=8)#------show only 8 characters data.
```

# - **Why use vertical=True?
# - It is useful when:**
- A DataFrame has many columns.
- Column values are very long.
- You want to inspect a single record clearly.
- Table output becomes difficult to read due to width.
