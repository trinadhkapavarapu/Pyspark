**🧠 When to Use to_json() in PySpark**
The to_json() function in PySpark is used to convert complex data types like StructType, MapType, or arrays into a JSON string. It’s the reverse of from_json().

**✅ Use to_json() When:**

- 1.	You want to serialize structured data (like a struct or map column) into a JSON string.
- 2.	You need to send data to external systems (e.g., Kafka, REST APIs, NoSQL databases) that expect JSON format.
- 3.	You’re preparing data for logging or exporting, where JSON is a preferred format.
- 4.	You want to flatten nested structures into a single string column for easier handling.


# Converting a Python dict to JSON

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_json
from pyspark.sql.types import StringType, MapType

# Create Spark session
spark = SparkSession.builder.getOrCreate()

# Sample data
data = [('trinadh', {'hair': 'black', 'eye': 'brown'})]
schema = ['name', 'properties']

# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show()
df.printSchema()


```

# Convert MapType column to JSON string

```python
# Convert MapType column to JSON string
df1 = df.withColumn('prop', to_json(df.properties))
df1.show()
df1.printSchema()
```

# Converting a StructType to JSON

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_json
from pyspark.sql.types import StringType, StructType, StructField

# Create Spark session
spark = SparkSession.builder.getOrCreate()

# Define schema
schema = StructType([
StructField('name', StringType()),
StructField('properties', StructType([
StructField('hair', StringType()),
StructField('eye', StringType())
]))
])

# Sample data
data = [('trinadh', ('black', 'brown'))]

# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show()
df.printSchema()


```

# Convert StructType column to JSON string

```python
# Convert StructType column to JSON string
df1 = df.withColumn('prop', to_json(df.properties))
df1.show()
df1.printSchema()
```
