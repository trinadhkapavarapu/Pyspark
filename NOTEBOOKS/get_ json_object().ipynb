**Definition - get_json_object()**
get_json_object() is a PySpark/SQL function used to extract a specific field or value from a JSON string column using a JSONPath expression, without requiring you to define a schema or parse the entire JSON structure.

**Use get_json_object() when:**
- •	You have a JSON string column and need only a few fields.
- •	The JSON contains nested objects, and you want to access values using JSONPath (for example, $.address.city).
- •	You need quick field extraction without converting the JSON into a struct using from_json().
- •	You are working with semi-structured data such as logs, Kafka messages, or API responses.


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import get_json_object
# Create Spark session
spark = SparkSession.builder.getOrCreate()
# Sample data with valid JSON strings
data = [
    ('tri', '{"address":{"city":"hyd","state":"telangana"},"gender":"male"}'),
    ('trinadh', '{"address":{"city":"kkd","state":"andhra"},"gender":"male"}')
]
schema = ['name', 'props']
# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show(truncate=False)
df.printSchema()

```

## Extract nested fields using `get_json_object`

```python
# Extract nested fields using get_json_object
df1 = df.select(
    'name',
    get_json_object('props', '$.gender').alias('gender'),
    get_json_object('props', '$.address.city').alias('city'),
    get_json_object('props', '$.address.state').alias('state')
)
df1.show(truncate=False)
df1.printSchema()
```


# note:-
json_tuple() and get_json_object() are both used to extract values from a JSON string. json_tuple() is used for extracting multiple top-level fields, whereas get_json_object() uses JSONPath and can extract both top-level and nested fields. Therefore, get_json_object() is more flexible for complex JSON structures.
