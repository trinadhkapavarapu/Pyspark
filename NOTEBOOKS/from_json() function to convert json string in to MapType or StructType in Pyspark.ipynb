**🧠 When to Use from_json() in PySpark**

The from_json() function is used to parse a JSON string column into a structured format like StructType or MapType. It’s powerful when working with semi-structured data that you want to convert into a schema-aware format for further processing.

**✅ Use from_json() When:**
- 1.	You have a JSON string column and want to convert it into a structured format (like a struct or map).
- 2.	You need to access multiple fields, especially nested ones, in a clean and schema-driven way.
- 3.	You want to perform transformations or filters on individual fields inside the JSON.
- 4.	You’re working with data from sources like Kafka, logs, or APIs that return JSON strings.


# convert json string in to MapType

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import from_json
from pyspark.sql.types import MapType, StringType
# Create Spark session
spark = SparkSession.builder.getOrCreate()
# Sample data
data = [('trinadh', '{"hair":"black","eye":"brown"}')]
schema = ['name', 'props']
# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show(truncate=False)
df.printSchema()

```

# Define MapType schema

```python
# Define MapType schema
mapTypeSchema = MapType(StringType(), StringType())
# Parse JSON string into MapType
df1 = df.withColumn('propsmap', from_json(df.props, mapTypeSchema))
df1.show(truncate=False)
df1.printSchema()
```

# Access individual keys from the map

```python
# Access individual keys from the map
df2 = df1.withColumn('hair', df1.propsmap['hair']) \
         .withColumn('eye', df1.propsmap['eye'])
df2.show(truncate=False)
```

# convert json string in to  StructType 

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import from_json
from pyspark.sql.types import StructType, StructField, StringType
# Create Spark session
spark = SparkSession.builder.getOrCreate()
# Sample data
data = [('trinadh', '{"hair":"black","eye":"brown"}')]
schema = ['name', 'props']
# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show(truncate=False)
df.printSchema()

```

# Define StructType schema

```python
# Define StructType schema
structTypeSchema = StructType([
    StructField('hair', StringType()),
    StructField('eye', StringType())
])
# Parse JSON string into StructType
df1 = df.withColumn('propsStruct', from_json(df.props, structTypeSchema))
df1.show(truncate=False)
df1.printSchema()
```

# Access individual fields from the struct

```python
# Access individual fields from the struct
df2 = df1.withColumn('hair', df1.propsStruct.hair) \
         .withColumn('eye', df1.propsStruct.eye)
df2.show(truncate=False)
```
