**🔄 Pivot in PySpark**

**Purpose: Converts rows into columns.**

- Use Case: When you want to aggregate data and display values of a column as new columns.

Or
- •	The pivot() function is used to rotate data in one column into multiple columns.
- •	It is an aggregation where one of the grouping column values will be converted into individual columns.

Or

- is used to transpose the list of values of a column to columns.

**# 🔁 Unpivot in PySpark**

****Purpose: Converts columns into rows.**

- Use Case: When you want to normalize wide data into a long format.

Or 
- is quite opposite to pivot i.e. transposing the columns into list of values to a column.

 Or

- •	Definition: Unpivoting is rotating columns into rows.
- •	Note: PySpark SQL does not have an UNPIVOT function, so we use the stack() function.


```python

# Sample data
data = [
    ("ABC","Q1",2000),
    ("ABC","Q2",3000),
    ("ABC","Q3",6000),
    ("ABC","Q4",1000),
    ("XYZ","Q1",5000),
    ("XYZ","Q2",4000),
    ("XYZ","Q3",1000),
    ("XYZ","Q4",2000),
    ("KLM","Q1",2000),
    ("KLM","Q2",3000),
    ("KLM","Q3",1000),
    ("KLM","Q4",5000)
]
columns = ["Company", "Quarter", "Revenue"]
# Create DataFrame
DF = spark.createDataFrame(data=data, schema=columns)
display(DF)


```

```python
# ----- PIVOT -----
Pivot_DF = DF.groupBy("Company").pivot("Quarter").sum("Revenue")
display(Pivot_DF)   # wide format with Q1..Q4 as columns
```

```python
# ----- UNPIVOT (back to long) -----
DF = Pivot_DF.selectExpr(
    "Company",
    "stack(4,'Q1',Q1,'Q2',Q2,'Q3',Q3,'Q4',Q4) as (Quarter, Revenue)"
)
display(DF)         # unpivoted long format
```
