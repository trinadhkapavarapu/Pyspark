_**What is join() in Python?**_
- join() is a string method in Python.
- It is used to combine all elements of an iterable (such as a list, tuple, or set) into a single string.
- A specified separator is placed between the elements.
- All elements must be strings; otherwise, they must be converted to strings first.
- separator.join(iterable)

```python
employee_data = [
    (10, "Michael Robinson", "1999-06-01", "100", 2000),
    (20, "James Wood", "2003-03-01", "200", 8000),
    (30, "Chris Andrews", "2005-04-01", "100", 6000),
    (40, "Mark Bond", "2008-10-01", "100", 7000),
    (50, "Steve Watson", "1996-02-01", "400", 1000),
    (60, "Mathews Simon", "1998-11-01", "500", 5000),
    (70, "Peter Paul", "2011-04-01", "600", 5000)
]
employee_schema = ["employee_id", "Name", "doj", "employee_dept_id", "salary"]
empDF = spark.createDataFrame(data=employee_data, schema=employee_schema)
display(empDF)

```

```python

#create iterable data 
column_list = empDF.columns
print(column_list)
#output
#['employee_id', 'Name', 'doj', 'employee_dept_id', 'salary']

#join the string
joined_string = ",".join(column_list)
print(joined_string)
#output
#employee_id,Name,doj,employee_dept_id,salary
```

```python
from pyspark.sql.functions import col
#Use cases
#1.Join Condition String/Generate JOIN Conditions Automatically
joinstring = " AND ".join(
    [f"Target.{col} = Source.{col}" for col in column_list]
)
print(joinstring)

#Output:
#Target.employee_id = Source.employee_id AND Target.Name = Source.Name AND Target.doj = Source.doj AND Target.employee_dept_id = Source.employee_dept_id AND Target.salary = Source.salary

#Use case: For MERGE or UPSERT operations where you need to match rows based on multiple columns.
```

```python
#2.Update Statement String/ Generate UPDATE SET Clause
updatestring = ",".join(list(map(lambda x: ("Target." + x + " = Source." + x), column_list)))
print(updatestring)


#Output:
#Target.employee_id = Source.employee_id,Target.Name=Source.Name,Target.doj=Source.doj,Target.employee_dept_id=Source.employee_dept_id,Target.salary=Source.salary

#Use case: For updating target table rows with source table values.
```

```python
#3.Source Insert String/Generate INSERT Column List
sourceinsertstring = ",".join(list(map(lambda x: ("Source." + x), column_list)))
print(sourceinsertstring)


#Output:
#Source.employee_id,Source.Name,Source.doj,Source.employee_dept_id,Source.salary

#Use case: For inserting new rows from source to target.

#Used for SQL INSERT statements.
```
