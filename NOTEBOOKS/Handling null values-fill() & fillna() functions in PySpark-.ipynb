**Main points to remember**
- Filter all records with null values
- Filter all records without null values
- Drop the records with null value-any
- Drop the records with null value-all
- Drop the records with null value on selected column
- fill value for all columns if null is present
- Fill value for specific columns if if contains null


```python
#Create dataframe

# Create DataFrame
data_student = [
    ("Michael", "M", "Science", "IT", 80, "P", 90),
    ("Nancy", "F", "Mathematics", None, 90, "P", None),
    ("David", "M", "English", "HR", 20, "F", 80),
    ("John", None, "Science", "IT", None, "F", None),
    ("Blessy", "F", None, None, 30, "F", 50),
    ("Martin", None, "Mathematics", "Finance", None, None, 70)
]

schema = ["name", "gender", "Subject", "dept", "Mark", "Status", "Attendance"]

df = spark.createDataFrame(data_student, schema)

display(df)

```

```python
#Filter all records with null values
# Method 1
# display(df.filter(df.Mark.isNull()))
# Method 2
from pyspark.sql.functions import col
display(df.filter(col("Mark").isNull()))
```

```python
#Filter all records without null values
display(df.filter(df.Mark.isNotNull()))

display(df.filter((df.Mark.isNotNull()) & (df.Attendance.isNotNull())))

display(df.filter((df.Mark.isNotNull()) | (df.Attendance.isNotNull())))
```

```python

#Drop the records with null value-any
display(df.na.drop())# ------------by default take (any)------removes one columns contains null
#display(df.dropna("any"))


```

```python
#Drop the records with null value-all
display(df.na.drop("all"))#----------removes which row contains all nulls
```

```python

#Drop the records with null value on selected column

display(df.na.drop(subset=["Mark"]))  # it could be combination of columns
display(df.na.drop(subset=["Mark", "Attendance"]))  # it could be combination of columns
```

```python
#fill value for all columns if null is present
df2 = (df.na.fill(0)          # numeric columns
         .na.fill("NA"))      # string columns
#display(df.fillna(0))
display(df2)


#or


df2 = (df.na.fill(value=0)          # numeric columns
         .na.fill(value="NA"))      # string columns
#display(df.fillna(value=0))
display(df2)
```

```python
#Fill value fro specific columns if if contains null

df.na.fill('unknown',['gender']).show()#---------------------------------single column
df.fillna('unknown',['dept']).show()
```

```python
df.na.fill('unknown',['gender','dept']).show()#--------------------------multiple column
```

```python
display(df.na.fill(value=0, subset=["Mark", "Attendance"]))#multiple column
display(df.na.fill({"Mark": 0, "Status": "NA", "Name": "No_name", "Subject": "English", "Attendance": 50}))
```
