- date_add() – Returns a new date by adding a specified number of days to a given date.
- date_sub() – Returns a new date by subtracting a specified number of days from a given date.
- datediff() – Returns the number of days between two dates.
- year() – Returns the year component of a date.
- dayofyear() – Returns the day of the year from the input date.
- month() – Returns the month component of a date.
- hour() – Returns the hour component of a timestamp.
- minute() – Returns the minute component of a timestamp.
- second() – Returns the second component of a timestamp.
- dayofmonth() – Returns the day of the month from the input date.
- add_months() – Returns a new date by adding a specified number of months to a given date.
- sub_month() – Returns a new date by subtracting a specified number of months from a given date.

```python
from pyspark.sql.functions import (
    date_add, date_sub, datediff, year, dayofyear, month,
    hour, minute, second, dayofmonth, add_months
)
```

```python
# Sample data
data = [("2022-03-15", "2022-03-16 12:34:56.789"),
        ("2022-03-01", "2022-03-16 01:23:45.678")]
# Create DataFrame
df = spark.createDataFrame(data, ["date", "timestamp"])
# Apply date functions
df_transformed = df \
    .withColumn("date_add_5", date_add("date", 5)) \
    .withColumn("date_sub_5", date_sub("date", 5)) \
    .withColumn("datediff", datediff("timestamp", "date")) \
    .withColumn("year", year("date")) \
    .withColumn("day_of_year", dayofyear("date")) \
    .withColumn("month", month("date")) \
    .withColumn("hour", hour("timestamp")) \
    .withColumn("minute", minute("timestamp")) \
    .withColumn("second", second("timestamp")) \
    .withColumn("day_of_month", dayofmonth("date")) \
    .withColumn("add_month_2", add_months("date", 2)) \
    .withColumn("sub_month_2", add_months("date", -2))
df_transformed.show(truncate=False)
```
