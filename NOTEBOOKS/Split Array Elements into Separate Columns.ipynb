```python
#Create the Initial DataFrame
df = spark.createDataFrame(
    [
        ['ABC', [1, 2, 3]],
        ['XYZ', [2, None, 4]],
        ['KLM', [8, 7]],
        ['IJK', [5]]
    ],
    ["key", "value"]
)
df.display()
```

#Split Array Values into Separate Columns

```python
spark.conf.set("spark.sql.ansi.enabled", "false")
```

```python
#Split Array Values into Separate Columns
df.select("key", df.value[0], df.value[1], df.value[2]).display()
```

#Count Number of Elements in Each Array

```python
#Count Number of Elements in Each Array
from pyspark.sql.functions import size
dfSize = df.select("key", "value", size("value").alias("NoOfArrayElements"))
dfSize.display()

```

#Get the Maximum Array Size

```python
from pyspark.sql.functions import *
#Get the Maximum Array Size
max_value = dfSize.agg({"NoOfArrayElements": "max"}).collect()[0][0]
print(max_value)
```

```python
spark.conf.set("spark.sql.ansi.enabled", "false")
```

#Split Array Elements into Separate Columns Dynamically

```python
#Split Array Elements into Separate Columns Dynamically
from pyspark.sql.functions import col

def arraySplitIntoCols(df, max_value):
    for i in range(int(max_value)):
        df = df.withColumn(f"new_col_{i}", col("value")[i])
    return df
dfExpanded = arraySplitIntoCols(df, max_value)
dfExpanded.display()


```
